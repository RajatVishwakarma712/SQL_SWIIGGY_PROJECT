```python
import matplotlib.pyplot as plt
```

plote the number of movies released per year (line chart)

```python
mnp=df.groupby('year')['id'].count()
```

```python
plt.figure(figsize=(10,6))
plt.plot(mnp.index,mnp.values,color="r")
plt.xlabel('year',color="r")
plt.ylabel('number of movies',color="purple")
plt.grid(color="skyblue")
plt.show()
```

plot the average rating over the years (line chart)

```python
roy=df.groupby('year')['rating'].mean()
```

```python
plt.figure(figsize=(10,6))
plt.plot(roy.index , roy.values , color = "g",marker="+")
plt.xlabel('year',color="r")
plt.ylabel('average rating',color="purple")
plt.grid(alpha=0.5,color="skyblue")
plt.show()
```

plot the distribuation of movie rating(histogram using matplotlib)

```python
plt.figure(figsize=(10,6))
plt.hist(df['rating'],color="magenta",edgecolor="black",bins=25)
plt.xlabel('rating')
plt.ylabel('frequency')
plt.title('rating distribution',color="r")
plt.grid(alpha=0.1,color="g")
plt.show()
```

which of the top 10 most frequent actors ?(horizontal bar chart)

```python
fa=df["stars_name"].str.split(",").explode().value_counts().head(10)
plt.figure(figsize=(10,6))
plt.barh(fa.index,fa.values,color="g")
plt.xticks(rotation=45)
plt.show()
```

Plot the relation ship between year and rating(scatter plot)

```python
plt.scatter(df['year'],df['rating'],color="r",marker="+")
plt.xlabel('year')
plt.ylabel('rating')
plt.show()
```
